# Conclusão e recomendações estratégicas

Neste notebook, vou consolidar os principais achados das análises logística e comercial.

O objetivo é responder à pergunta final do case: com base nos atrasos, margens e eficiência comercial, onde a Yandeh deve concentrar seus esforços e recursos?

## Achados logísticos

A análise logística mostrou que o principal ponto de atenção está nas carretas.

`Carreta_Bau` e `Carreta_Sider` representam juntas cerca de 72% da margem bruta total da operação. Porém, essas mesmas frotas também apresentam as maiores probabilidades de atraso, acima de 64%, e os maiores atrasos médios, acima de 25 minutos.

Isso indica que as carretas são financeiramente críticas, mas também concentram maior risco operacional.

O VUC possui o maior volume de entregas, mas baixa participação na margem total e menor atraso médio. Já o Truck se destacou pela maior margem média por km, mas com menor participação financeira total em comparação às carretas.

## Achados comerciais

A análise comercial mostrou que `Field Sales` apresentou melhor desempenho nos principais indicadores avaliados.

Mesmo com menos vendedores e menos leads trabalhados, Field Sales teve maior taxa de conversão ponderada, maior faturamento estimado total, maior faturamento médio por vendedor e maior faturamento estimado por lead.

A taxa de conversão ponderada foi de 16,10% em Field Sales contra 9,42% em Inside Sales. O faturamento estimado por lead também foi maior em Field Sales, com R$ 204,52 contra R$ 112,05 em Inside Sales.

A experiência dos vendedores não apresentou relação positiva clara com desempenho, sugerindo que o canal comercial pode explicar melhor a diferença de performance do que os anos de experiência isoladamente.

## Recomendação estratégica

Com base nos dados analisados, a Yandeh deve priorizar esforços na melhoria operacional das frotas `Carreta_Bau` e `Carreta_Sider`. Essas frotas concentram cerca de 72% da margem bruta total, mas também apresentam as maiores probabilidades e intensidades de atraso. Portanto, o foco não deve ser reduzir sua participação, e sim melhorar previsibilidade, roteirização, planejamento de carga/descarga e controle de prazos nessas operações.

Na frente comercial, `Field Sales` apresentou melhor eficiência em conversão, faturamento estimado e faturamento por lead. Assim, a empresa deveria investigar quais práticas desse canal podem ser replicadas e avaliar sua expansão com cautela, considerando que o cálculo completo de ROI exigiria dados adicionais de custo de visita e operação.

## Limitações da análise

A margem calculada neste projeto considera apenas o valor do frete menos o custo de combustível. Portanto, ela deve ser interpretada como uma margem bruta simplificada, já que a base não inclui outros custos relevantes, como motorista, manutenção, pedágio, depreciação, seguro e custos administrativos.

Além disso, a análise comercial não possui dados de custo por canal, como custo de visita, salário, comissão ou custo de atendimento. Por isso, os resultados indicam eficiência comercial, mas não permitem calcular o ROI completo de cada canal.
